```python
from qiskit import QuantumCircuit, ClassicalRegister, QuantumRegister
from qiskit.visualization import plot_histogram
from qiskit_aer import AerSimulator
%matplotlib inline
```

```python
classicalRegister = ClassicalRegister(1)
quantumRegister = QuantumRegister(1)
circuit = QuantumCircuit(quantumRegister, classicalRegister)
circuit.measure(quantumRegister, classicalRegister)
circuit.draw(output='mpl');
```

```python
#Shorthand method
circuit2 = QuantumCircuit(1)
circuit2.measure_all()
print(circuit2)
```

```python
simulator = AerSimulator()
result = simulator.run(circuit, shots=1024).result() #runs simulation 1024x to get a probabilistic distribution
count = result.get_counts()
plot_histogram(count);
#defaults to initial value of 0
```

```python

circuit = QuantumCircuit(1)
circuit.h(0) #puts circuit in a state of superposition
circuit.measure_all()
simulator = AerSimulator()
result = simulator.run(circuit, shots=1024).result()
count = result.get_counts()
plot_histogram(count);
```
